```python
#largest number
def largest_number(l):
    large = l[0]
    for i in l:
        if i> large:
            large = i
    return large
```

```python
#second largest number
def second_largest_number(l):
    first = second = float('-inf')
    for i in l:
        if i>first:
            second = first
            first = i
        elif i > second and i != first:
            second = i
    return second
```

```python
def check_sorted_list(l):
    for i in l:
        if l[i+1]<l[i]:
            return True
    return False
```

```python
def remove_duplicates_from_sorted_list_bruteforce(l):
    unique = set()
    for i in l:
        unique.append(i)
    return list(unique)
```

```python
def unique_elements(arr):

    if not arr:
        return []

    result = [arr[0]]

    for i in range(1, len(arr)):
        if arr[i] != arr[i - 1]:
            result.append(arr[i])

    return result
```

```python
l = [1,1,2,2,2,3,3,4]

unique_elements(l)
```
