```python
# Print something N times using recursion

def print_n_times(n, text):
    
    # Base condition
    if n == 0:
        return
    
    print(text)
    
    # Recursive call
    print_n_times(n - 1, text)


# Taking input from user
n = int(input("Enter the number of times you want to print: "))
text = input("Enter the text you want to print: ")

# Function call
print_n_times(n, text)
```

```python
#print 1 to N series
def one_to_n(i,n):
    if i > n:
        return
    print(i)

    one_to_n(i+1,n)

one_to_n(1,10)
```

```python
#sum of first n numbers PARAMETERIZED WAY

def sum_n(i,total):
    if i<1:
        print(total)
        return
    sum_n(i-1,total + i)

sum_n(3,0)
```

```python
#sum of first n numbers
def sum_n_f(n):
    if n == 0:
        return 0
    return n+sum_n_f(n-1)

sum_n_f(3)
```

```python
def fact_n(n):
    if n == 0:
        return 1
    return n*fact_n(n-1)

fact_n(3)
```

```python
def fibonacci(n):
    if n<=1:
        return n
    return fibonacci(n-1) + fibonacci(n-1)
```

```python
def palindrome_check(s):

    # BASE CASE
    if len(s) <= 1:
        return True

    # NOT PALINDROME
    if s[0] != s[-1]:
        return False

    # SMALLER PROBLEM
    return palindrome_check(s[1:-1])


print(palindrome_check("madam"))
```
