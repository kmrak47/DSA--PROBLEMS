```python
#count all digits of a number
'''
Approach : 
1) declare var with 0 ; count.
2) run a loop 
3) check if the entry is number or not 
4) if number then add to count variable which we create in step 1
5) print result

TC - O(n) SC : O(1)
'''
def countallnos(x):
    cnt = 0
    while(x>0):
        lst_dg = x%10
        cnt = cnt + 1
        x=x//10
    return cnt

countallnos(156)
```

```python
#Reverse a number
'''
Approach: reverse_number = reverse_number *10 + last_digit

'''

def rev_num(x):
    rev_num = 0
    while(x>0):
        lst_dg = x%10
        rev_num = (rev_num * 10)+ lst_dg
        x=x//10
    return rev_num

rev_num(134)
```

```python
#check palindrome
'''
Approach: 1331 = 1331 ;its a plaindrome number

'''
def palindrome(x):
    n = x
    rev_num = 0
    while(n>0):
        last_digit = n%10
        rev_num = (rev_num * 10 )+ last_digit
        n=n//10
    
    return rev_num == x

palindrome(1201)
```

```python
#armsstrong number
'''
ex: 371 cube of every digit if summed up it gives same value.

'''

def armstrong(x):
    n=x
    total = 0
    while(x>0):
        last_digit = x%10
        total = total+(last_digit**3)
        x = x//10

    return total==n

armstrong(371)
```

```python
#print all divisors
def pod(x):
    l = []
    for i in range(1,int(x**0.5),1):
        if (x%i) == 0:
            l.append(i)
            if x//i != i:
                l.append(x//i)
    return sorted(l)

pod(136)
```

```python
#check prime number
def prime(x):
    cnt = 0
    for i in range(1,int(x**0.5),1):
        if x%i==0:
            cnt = cnt + 1
            if (x/i) != i:
                cnt = cnt+1
    if cnt == 2:
        return "prime"
    elif cnt > 2:
        return "not prime"
    


prime(136)
```

```python
#gcd
def gcd(a, b):
    
    # convert negative to positive
    a = abs(a)
    b = abs(b)
    
    # edge case
    if a == 0 and b == 0:
        return "GCD Not Defined"
    
    # Euclidean Algorithm
    while b != 0:
        a, b = b, a % b
        
    return a

gcd(20,10)
```
