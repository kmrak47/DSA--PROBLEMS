```python
#Selection Sort
def selection_sort(l):
    n = len(l)
    for i in range(n-1):
        min_index = i
        for j in range(i+1,n):
            if l[j]<l[min_index]:
                min_index = j
        l[i], l[min_index] = l[min_index],l[i]
    return l
```

```python
def bubble_sort(l):
    n = len(l)
    for i in range(n-1):
        swapped = False
        for j in range(n-i-1):
            if l[j] > l[j+1]:
                l[j], l[j+1] = l[j+1], l[j]
                swapped = True
        if not swapped:
            break
    return l
```
